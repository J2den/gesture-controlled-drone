```python
from djitellopy import Tello
import cv2, math, time
```

```python
tello = Tello()
```

```python

tello.connect()
```

```python
import numpy as np

# Create a dummy image
img = np.zeros((200, 200, 3), dtype=np.uint8)
cv2.imshow("Dummy", img)
```

```python
tello.streamon()
frame_read = tello.get_frame_read()
```

```python
tello.takeoff()
```

```python
while True:
    img = frame_read.frame
    cv2.imshow("drone", img)

    key = cv2.waitKey(1) & 0xff
    if key == 27:
        break
    elif key == ord('w'):
        tello.move_forward(20)
    elif key == ord('s'):
        tello.move_back(20)
    elif key == ord('a'):
        tello.move_left(20)
    elif key == ord('d'):
        tello.move_right(20)
    elif key == ord('e'):
        tello.rotate_clockwise(20)
    elif key == ord('q'):
        tello.rotate_counter_clockwise(20)
    elif key == ord('r'):
        tello.move_up(20)
    elif key == ord('f'): 
        tello.move_down(20)
        
tello.land()
```
